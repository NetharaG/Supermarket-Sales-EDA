## Tools

## 1. Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (9, 5)
```

## 2. Load the dataset
Upload `supermarket_sales.csv` to the Colab session when prompted.

```python
from google.colab import files
uploaded = files.upload()  # Select supermarket_sales.csv

df = pd.read_csv("supermarket_sales.csv")
df.head()
```

## 3. Understand the data

```python
print("Shape:", df.shape)
print("\\nColumn information:")
df.info()
print("\\nSummary statistics:")
display(df.describe(include="all").T)
print("\\nMissing values:")
display(df.isnull().sum())
print("\\nDuplicate rows:", df.duplicated().sum())
```

## 4. Data cleaning and preprocessing

```python
# Remove duplicate records
df = df.drop_duplicates().copy()

# Fill missing numerical values with median and categorical values with mode
for col in df.columns:
    if df[col].isnull().any():
        if pd.api.types.is_numeric_dtype(df[col]):
            df[col] = df[col].fillna(df[col].median())
        else:
            df[col] = df[col].fillna(df[col].mode()[0])

# Convert date to datetime and derive useful features
df["Date"] = pd.to_datetime(df["Date"])
df["Month"] = df["Date"].dt.month_name()
df["Day"] = df["Date"].dt.day_name()

print("Cleaned shape:", df.shape)
print("Missing values after cleaning:", df.isnull().sum().sum())
```

## 5. Exploratory Data Analysis

```python
print("Total revenue:", round(df["Total"].sum(), 2))
print("Average transaction:", round(df["Total"].mean(), 2))
print("Total transactions:", len(df))
print("Average customer rating:", round(df["Rating"].mean(), 2))
```

## 6. Visualizations

```python
# 1. Revenue by product line
plt.figure(figsize=(10,5))
sns.barplot(data=df, x="Total", y="Product_line", estimator=sum, errorbar=None)
plt.title("Total Revenue by Product Line")
plt.xlabel("Revenue"); plt.ylabel("Product line")
plt.show()
```

```python
# 2. Revenue by branch
plt.figure(figsize=(7,5))
sns.barplot(data=df, x="Branch", y="Total", estimator=sum, errorbar=None)
plt.title("Total Revenue by Branch")
plt.show()
```

```python
# 3. Payment method distribution
plt.figure(figsize=(7,5))
sns.countplot(data=df, x="Payment", order=df["Payment"].value_counts().index)
plt.title("Transactions by Payment Method")
plt.xlabel("Payment method"); plt.ylabel("Number of transactions")
plt.show()
```

```python
# 4. Distribution of transaction totals
plt.figure(figsize=(8,5))
sns.histplot(df["Total"], bins=25, kde=True)
plt.title("Distribution of Transaction Values")
plt.xlabel("Transaction total")
plt.show()
```

```python
# 5. Quantity vs total
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Quantity", y="Total", hue="Customer_type", alpha=.7)
plt.title("Quantity vs Transaction Total")
plt.show()
```

```python
# 6. Correlation heatmap
numeric_cols = ["Unit_price", "Quantity", "Tax_5_percent", "Total", "COGS", "Gross_income", "Rating"]
plt.figure(figsize=(9,6))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

## 7. Key findings (derive from your output)

```python
print("Highest revenue product line:")
display(df.groupby("Product_line")["Total"].sum().sort_values(ascending=False).head(1))
print("Highest revenue branch:")
display(df.groupby("Branch")["Total"].sum().sort_values(ascending=False).head(1))
print("Most used payment method:")
display(df["Payment"].value_counts().head(1))
print("Average rating by product line:")
display(df.groupby("Product_line")["Rating"].mean().sort_values(ascending=False))
```
